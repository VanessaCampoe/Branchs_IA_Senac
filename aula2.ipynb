```python
# crie um grafico de linha tempo vs retorno
import matplotlib.pyplot as plt

# 2 - Selecionar o vetor (array) de preços
precos = df[0].values   # <-- nome numérico da coluna
# print("Preços:", precos)

# 3 - Calcular o retorno financeiro
ret = (precos[1:] - precos[:-1]) / precos[:-1]
# print("Retornos:", ret)

# 4 - Plotar
plt.figure(figsize=(8,4))
plt.plot(ret, marker="o")
plt.title("Retorno Financeiro (ret)")
plt.xlabel("Período")
plt.ylabel("Retorno")
plt.grid(True)
plt.show()
```
